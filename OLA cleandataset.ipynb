```python
#null values fill krna#
df["V_TAT"]=df["V_TAT"].fillna(df["V_TAT"].mean())
df["C_TAT"]=df["C_TAT"].fillna(df["C_TAT"].mean())
df["Canceled_Rides_by_Customer"]=df["Canceled_Rides_by_Customer"].fillna(df["Canceled_Rides_by_Customer"].mode()[0])
df["Canceled_Rides_by_Driver"]=df["Canceled_Rides_by_Driver"].fillna(df["Canceled_Rides_by_Driver"].mode()[0])
df["Incomplete_Rides"]=df["Incomplete_Rides"].fillna(df["Incomplete_Rides"].mode()[0])
df["Incomplete_Rides_Reason"]=df["Incomplete_Rides_Reason"].fillna(df["Incomplete_Rides_Reason"].mode()[0])
df["Payment_Method"]=df["Payment_Method"].fillna(df["Payment_Method"].mode()[0])
df["Driver_Ratings"]=df["Driver_Ratings"].fillna(df["Driver_Ratings"].mean())
df["Customer_Rating"]=df["Customer_Rating"].fillna(df["Customer_Rating"].mean())
```

```python
df.isnull().sum()
```

```python
df[df["Booking_Status"]=="Success"]
```

```python
 #Find the average ride distance for each vehicle type:
avg_distance = df.groupby("Vehicle_Type")["Ride_Distance"].mean()
```

```python
print(avg_distance)
```

```python
#. Get the total number of cancelled rides by customers:
total_cancelled = (df["Booking_Status"] == "Canceled_Rides_by_Customer").sum()

print("\nTotal Cancelled Rides by Customers:", total_cancelled)
```

```python
 #List the top 5 customers who booked the highest number of rides:
# Top 5 customers who booked the highest number of rides

top_5_customers = (df.groupby("Customer_ID").size().reset_index(name="Total_Rides").sort_values("Total_Rides", ascending=False).head(5))

print(top_5_customers)
```

```python
top_10_customer =(df.groupby("Customer_ID").size().reset_index(name="total_rides").sort_values("total_rides",ascending=False).head(10))
print(top_10_customer)
```

```python
#Get the number of rides cancelled by drivers due to personal and car-related issues
rides_cancelled = (df["Canceled_Rides_by_Driver"]=="Personal & Car related issue").sum()
print(" the number of rides cancelled by drivers due to personal and car-related issues:",rides_cancelled)
```

```python
#Find the maximum and minimum driver ratings for Prime Sedan bookings:
# Filter Prime Sedan bookings
prime_sedan = df[df["Vehicle_Type"] == "Prime Sedan"]

# Maximum and Minimum Driver Rating
max_rating = prime_sedan["Driver_Ratings"].max()
min_rating = prime_sedan["Driver_Ratings"].min()

print("Maximum Driver Rating:", max_rating)
print("Minimum Driver Rating:", min_rating)
```

```python
#. Retrieve all rides where payment was made using UPI:
total_UPI = (df["Payment_Method"] == "UPI").sum()

print("\nTotal Numbers of UPI pay:", total_UPI)
```

```python
#. Find the average customer rating per vehicle type
avg_cutomer_rating = df.groupby("Vehicle_Type")["Customer_Rating"].mean()
print("avg_cutomer_rating",avg_cutomer_rating )
```

```python
#Calculate the total booking value of rides completed successfully
total_successful_booking=(df["Booking_Status"]=="Success").sum()
print("\ntotal_successful_booking:",total_successful_booking)
```

```python
#List all incomplete rides along with the reason
incomplete_rides=(df["Incomplete_Rides"]=="Yes").sum()
print(incomplete_rides)
```
